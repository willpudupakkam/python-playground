```python
import random
import time
import matplotlib.pyplot as plt

with open('./words.txt', 'r') as myfile:
     raw_text = myfile.read()
     wordlist = raw_text.split('\n')
     random.shuffle(wordlist)
     
def met_conditions(guess):
    global secret_word
    global learned
    global exists
    global not_exists
    global already_guessed
    
    if guess in already_guessed:
         return False
    
    for i, char in enumerate(guess):
         if char in not_exists or (learned[i] != '' and char != learned[i]):
              return False
         
    for letter in exists:
         if letter not in guess:
              return False
         
    return True

def guess_word(guess):
    global secret_word
    global learned
    global exists
    global not_exists
    global wordlist
    global count

#   print(guess)
    already_guessed.append(guess)
    count += 1

    # Did we win?
    if guess == secret_word:
     #     time.sleep(0.25)
         return #print("=== Win ===")

    # No we didn't, so we learn from mistakes
    for i, char in enumerate(guess):
            if char == secret_word[i]:
                learned[i] = char
                exists.append(char)
            elif char in secret_word:
                exists.append(char)
            else:
                not_exists.append(char)

    # Use new knowledge to justify new word choice
    for word in wordlist:
        if met_conditions(word):
             return guess_word(word)
    
    return print("Secret word not in wordlist")
  
data = {}
mean_data = {}
mean = 0
count = 0
num_of_trials = 2500

for word in random.sample(wordlist, 10):
    for i in range(num_of_trials):  
        secret_word = wordlist[random.randint(0, len(wordlist)-1)]  # Random secret word
        #print(f'Secret word is \'{secret_word}\'')
        learned = {0:'', 1:'', 2:'', 3:'', 4:''}
        exists = []
        not_exists = []
        already_guessed = []
        random_guess = word #wordlist[random.randint(0, len(wordlist)-1)]     
        guess_word(random_guess)
        if count in data:
            data[count] += 1
        else:
            data[count] = 1
        count = 0
    print(f'Tried \'{random_guess}\'')
    sorted_data = dict(sorted(data.items()))
    for key in sorted_data:
        mean += (sorted_data[key] * key) / num_of_trials
    mean_data[word] = mean
    mean = 0
    data = {}

best_word = min(mean_data, key=mean_data.get)
print(f'The best starting word was {best_word}, with a mean of {mean_data[best_word]}')
# def plot_hist_from_counts(counts, xlabel="Value", ylabel="Count", title="Histogram"):
#     xs = sorted(counts.keys())          # left→right order
#     ys = [counts[k] for k in xs]

#     plt.figure()
#     plt.bar(xs, ys, width=0.8, align="center")
#     plt.xlabel(xlabel)
#     plt.ylabel(ylabel)
#     plt.title(title)
#     plt.xticks(xs)
#     plt.tight_layout()
#     plt.show()

# plot_hist_from_counts(sorted_data)
```
